# Análise Estatística

## Objetivo

Avaliar estatisticamente os principais padrões identificados durante a Análise Exploratória de Dados (EDA), verificando se as alterações observadas entre maio e junho de 2018 apresentam evidências estatísticas consistentes.

A análise será conduzida a partir das hipóteses formuladas durante a EDA, respeitando a estrutura dos dados e os pressupostos dos métodos estatísticos utilizados.

## Fenômeno investigado

A análise estatística parte da deterioração recente identificada entre maio e junho de 2018.

Nesse período, observou-se:

- redução de 13,19% na receita de produtos;
- redução de 10,11% no número de pedidos;
- redução de 3,42% no ticket médio.

A EDA aprofundou esse fenômeno e identificou que a deterioração não ocorreu de maneira uniforme entre os produtos e categorias.

Entre as dez categorias prioritárias analisadas, foram identificados 2.413 produtos com participação nas vendas do período. Dentro desse conjunto, 104 produtos concentraram aproximadamente 50% das perdas de receita observadas entre os produtos que apresentaram queda.

A investigação exploratória também mostrou que, entre esses 104 produtos de maior impacto, a redução do número de pedidos foi o principal padrão associado à perda de receita observada.

Esses resultados são utilizados como ponto de partida para a formulação das hipóteses estatísticas. Eles representam achados exploratórios e não constituem, por si só, evidência de causalidade.

## Pergunta estatística central

A principal questão estatística desta etapa é:

> **Os indicadores de desempenho dos produtos apresentaram alterações sistemáticas entre maio e junho de 2018, além das variações que poderiam ser atribuídas ao comportamento observado nas unidades analisadas?**

A investigação será inicialmente concentrada em três dimensões:

1. **Número de pedidos por produto**, para avaliar se houve alteração sistemática no volume de pedidos;
2. **Ticket médio por produto**, para avaliar se houve alteração no valor médio associado aos pedidos dos produtos que permaneceram ativos nos dois períodos;
3. **Status de atividade dos produtos**, para avaliar se houve alteração na proporção de produtos que registraram vendas entre os dois meses.

A escolha dos métodos estatísticos será realizada posteriormente, após a construção das bases analíticas e o diagnóstico das características das variáveis.

A pergunta estatística não busca demonstrar causalidade. Seu objetivo é avaliar a existência, magnitude e consistência das alterações observadas entre os dois períodos.

## Escopo e população analítica

A análise estatística será realizada sobre os produtos pertencentes às dez categorias prioritárias identificadas durante a EDA.

Esse conjunto contém 2.413 produtos com participação nas vendas das categorias selecionadas durante o período analisado.

A definição dessa população busca evitar que a inferência estatística seja realizada exclusivamente sobre os 104 produtos de maior impacto econômico, uma vez que esses produtos foram selecionados com base na própria magnitude da perda observada.

Os 104 produtos serão mantidos como um grupo de interesse para análises descritivas e de impacto econômico, mas não serão utilizados isoladamente como população de referência para testar a própria ocorrência da deterioração.

### População analítica principal

**2.413 produtos pertencentes às dez categorias prioritárias.**

### Grupo de maior impacto

**104 produtos que concentram aproximadamente 50% das perdas observadas entre os produtos com queda de receita nesse conjunto.**

### Períodos comparados

- Maio de 2018;
- Junho de 2018.

A estrutura temporal permite comparar as mesmas unidades de análise entre dois períodos consecutivos, possibilitando avaliar posteriormente se a comparação deve ser tratada como pareada.

### Unidade de análise

A unidade estatística principal será o **produto (`product_id`)**.

A população analítica foi definida antes da aplicação dos testes estatísticos. ## Princípios metodológicos

A análise estatística seguirá os seguintes princípios:

### 1. Método orientado pela estrutura dos dados

A escolha dos métodos estatísticos será determinada pela natureza das variáveis, pela unidade de análise, pela estrutura temporal e pelas características observadas nos dados.

Nenhum teste será escolhido previamente apenas por conveniência ou para produzir determinado resultado.

### 2. Hipóteses definidas antes dos testes

Os padrões identificados durante a EDA serão transformados em hipóteses estatísticas formalmente definidas antes da execução dos testes.

As hipóteses nula e alternativa serão estabelecidas de acordo com cada questão analisada.

### 3. Avaliação dos pressupostos

Antes da aplicação de qualquer teste, serão avaliadas as características relevantes das variáveis e dos dados, incluindo distribuição, assimetria, valores extremos, zeros, empates, dependência entre observações e estrutura pareada quando aplicável.

### 4. Significância estatística não implica relevância de negócio

Um resultado estatisticamente significativo não será interpretado automaticamente como uma alteração relevante para o negócio.

Sempre que apropriado, os resultados serão acompanhados por medidas de tamanho de efeito e intervalos de confiança.

### 5. Ausência de causalidade automática

Os testes estatísticos serão utilizados para avaliar diferenças, associações e padrões nos dados.

Os resultados não serão interpretados como evidência de causalidade sem um desenho de estudo que permita sustentar esse tipo de conclusão.

### 6. Preservação da estrutura dos dados

Valores ausentes, zeros e observações sem definição matemática não serão convertidos ou imputados arbitrariamente.

Qualquer tratamento necessário será definido de acordo com o significado da variável e documentado antes da análise.

### 7. Separação entre análise inferencial e análise de impacto

Os 2.413 produtos da população analítica serão utilizados como referência para a análise estatística principal.

Os 104 produtos de maior impacto serão utilizados como grupo de interesse para compreender a concentração e a magnitude econômica da deterioração.

Essa distinção evita confundir seleção por impacto com evidência inferencial sobre a ocorrência da própria queda.

### 8. Transparência e reprodutibilidade

Todas as decisões metodológicas, transformações de dados, critérios de inclusão e exclusão, testes aplicados e interpretações serão documentados no notebook.

O objetivo é permitir que outra pessoa consiga compreender e reproduzir o processo analítico.

## Limitações e cuidados da análise estatística

A análise estatística deve ser interpretada considerando as características e limitações do conjunto de dados.

### Comparação temporal

A análise principal compara maio e junho de 2018, dois meses completos e consecutivos.

Essa comparação permite avaliar alterações entre os períodos, mas não é suficiente, isoladamente, para caracterizar uma tendência estrutural de longo prazo.

### Natureza observacional dos dados

Os dados utilizados são observacionais. Não houve controle experimental sobre fatores que podem influenciar o desempenho comercial.

Consequentemente, diferenças estatisticamente significativas entre os períodos não serão interpretadas automaticamente como relações causais.

### Distribuição das variáveis

O número de pedidos por produto é uma variável discreta e pode apresentar assimetria, concentração de valores baixos e grande quantidade de zeros.

Essas características deverão ser avaliadas antes da escolha dos testes estatísticos.

### Produtos sem pedidos

O valor zero possui significado válido para a variável número de pedidos.

Entretanto, quando um produto não apresenta pedidos em determinado período, seu ticket médio não é definido, pois a divisão entre receita e número de pedidos resulta em uma operação do tipo 0/0.

Essas observações não serão convertidas arbitrariamente em ticket igual a zero.

### Heterogeneidade entre produtos

Os produtos apresentam diferentes níveis de volume de vendas e diferentes valores monetários.

Por esse motivo, médias agregadas não serão utilizadas isoladamente para representar o comportamento da população analítica.

Sempre que necessário, serão consideradas medidas de distribuição, diferenças individuais e medidas de efeito.

### Seleção dos produtos de maior impacto

Os 104 produtos de maior impacto foram identificados com base na própria variação de receita observada entre maio e junho de 2018.

Por esse motivo, esse grupo não será tratado isoladamente como uma amostra independente para testar a existência da queda que determinou sua seleção.

Sua utilização principal será descritiva e diagnóstica.

### Interpretação dos resultados

Os resultados estatísticos serão interpretados em conjunto com:

- magnitude da diferença;
- tamanho do efeito;
- intervalo de confiança;
- características da distribuição;
- relevância para o problema de negócio.

O valor de p não será utilizado isoladamente como critério para determinar a importância de um resultado.

### Causalidade

A análise estatística poderá fornecer evidências sobre diferenças ou associações observadas nos dados.

Ela não permitirá, por si só, concluir que determinado produto, categoria ou seller causou a deterioração observada.

Qualquer afirmação causal deverá exigir evidências adicionais e será tratada separadamente na etapa de diagnóstico.

# Parte 2 — Definição da Unidade Estatística

## 2.1 Unidade estatística

A unidade estatística principal desta análise será o **produto (`product_id`)**.

Essa definição decorre da pergunta estatística estabelecida na etapa anterior: avaliar se os indicadores de desempenho apresentaram alterações sistemáticas entre maio e junho de 2018.

Embora o pedido seja a principal unidade de negócio da operação e a linha de item represente a ocorrência de um produto dentro de um pedido, a análise estatística desta etapa precisa acompanhar o desempenho das mesmas unidades entre os dois períodos.

Dessa forma:

- `order_id` representa o pedido;
- `(order_id, order_item_id)` representa uma linha de item;
- `product_id` representa a unidade estatística principal da análise.

Cada produto será observado em dois períodos:

- maio de 2018;
- junho de 2018.

A utilização do `product_id` permite construir comparações intraunidade, nas quais o comportamento de cada produto em junho é comparado ao seu próprio comportamento em maio.

Essa estrutura será considerada posteriormente na definição dos métodos estatísticos apropriados.

## 2.2 População analítica

A população analítica principal será composta pelos **2.413 produtos (`product_id`) pertencentes às dez categorias prioritárias identificadas durante a EDA**.

As categorias foram definidas anteriormente com base nas maiores perdas absolutas de receita entre maio e junho de 2018.

A população de 2.413 produtos será utilizada como referência para a análise estatística das alterações de desempenho entre os dois períodos.

Os **104 produtos de maior impacto econômico** permanecerão como um grupo de interesse para análises descritivas e de diagnóstico.

Essa distinção é necessária porque os 104 produtos foram selecionados com base na própria variação de receita observada. Utilizá-los isoladamente como população para testar a existência da queda introduziria uma seleção condicionada ao próprio resultado analisado.

Portanto:

| Conjunto | Função na análise |
|---|---|
| 2.413 produtos | População analítica principal |
| 104 produtos | Grupo de maior impacto para análise descritiva e diagnóstico |

A população analítica será definida antes da aplicação dos testes estatísticos, evitando que a seleção das observações seja determinada pelos resultados dos testes.

## 2.3 Estrutura temporal e pareamento

A análise compara as mesmas unidades estatísticas em dois períodos consecutivos:

- maio de 2018;
- junho de 2018.

Como cada `product_id` pode ser identificado nos dois períodos, existe uma correspondência natural entre as observações de maio e junho.

A estrutura pode ser representada como:

| Produto | Maio | Junho |
|---|---:|---:|
| Produto A | valor em maio | valor em junho |
| Produto B | valor em maio | valor em junho |
| Produto C | valor em maio | valor em junho |

Dessa forma, a comparação principal possui estrutura **pareada**, pois cada produto é comparado com ele mesmo ao longo dos dois períodos.

Para uma variável quantitativa, a diferença individual poderá ser representada por:

**Δᵢ = valorᵢ,junho − valorᵢ,maio**

Essa estrutura será importante para a escolha dos métodos estatísticos, pois a análise deverá considerar a dependência existente entre as observações correspondentes de cada produto.

A classificação como estrutura pareada não implica, por si só, a escolha de um teste específico. A definição do método será realizada posteriormente após o diagnóstico das variáveis e de seus pressupostos.

## 2.4 Variáveis estatísticas

A análise estatística utilizará inicialmente três dimensões principais de desempenho por produto.

### 1. Número de pedidos

Representa a quantidade de pedidos distintos nos quais o produto esteve presente durante cada mês.

Variáveis:

- `pedidos_maio`;
- `pedidos_junho`.

A diferença entre os períodos será representada por:

**Δ pedidos = pedidos_junho − pedidos_maio**

O valor zero é válido para essa variável e representa ausência de pedidos no período.

### 2. Ticket médio por produto

Representa a receita de produtos dividida pelo número de pedidos distintos associados ao produto no período.

Variáveis:

- `ticket_maio`;
- `ticket_junho`.

A definição será:

**ticket = receita do produto / número de pedidos do produto**

Quando um produto não apresentar pedidos em determinado período, o ticket correspondente não será definido, pois a divisão resulta em uma operação do tipo 0/0.

Essas observações não serão convertidas arbitrariamente para zero.

### 3. Status de atividade

Representa se o produto registrou pelo menos um pedido no período.

A variável será definida como:

- `1` = produto apresentou pelo menos um pedido;
- `0` = produto não apresentou pedidos.

Variáveis:

- `ativo_maio`;
- `ativo_junho`.

Essa dimensão permitirá avaliar separadamente a transição entre atividade e ausência de atividade dos produtos.

### Variável complementar

A receita por produto poderá ser utilizada como variável descritiva e de apoio ao diagnóstico, mas a análise estatística inicial será concentrada nos componentes de volume, ticket e atividade identificados durante a EDA.

## 2.5 Critérios de inclusão e exclusão

A construção da população estatística seguirá critérios explícitos de inclusão e tratamento.

### Inclusão

Serão incluídos na população analítica:

- produtos pertencentes às dez categorias prioritárias definidas na EDA;
- produtos com `product_id` válido;
- produtos observados na base de vendas durante o período analisado;
- observações referentes a maio e junho de 2018.

### Tratamento de ausência de pedidos

A ausência de pedidos em determinado mês será representada por zero para a variável de número de pedidos.

Isso preserva o significado operacional da informação:

**zero pedidos = produto não registrou pedidos no período.**

### Tratamento do ticket

Quando o número de pedidos for zero, o ticket médio não será definido.

Essas observações não serão convertidas automaticamente para zero, pois zero pedidos não implica ticket igual a zero.

O tratamento dessas observações dependerá da hipótese estatística analisada.

### Exclusão

Não serão excluídas automaticamente observações apenas por apresentarem:

- zero pedidos;
- ticket indefinido decorrente de zero pedidos;
- valores extremos;
- grande variação entre os períodos.

Eventuais exclusões ou filtros adicionais somente serão realizados quando houver justificativa metodológica específica e serão documentados no notebook.

### Grupo de maior impacto

Os 104 produtos de maior impacto não serão utilizados como população estatística principal para testar a própria queda que determinou sua seleção.

Eles permanecerão disponíveis para análises descritivas e de diagnóstico.

# Parte 3 — Construção da Base Estatística

## Objetivo

Construir uma base analítica específica para a análise estatística, utilizando como unidade de observação o `product_id` e organizando as informações de maio e junho de 2018 de forma comparável.

A base será construída a partir das tabelas `orders`, `order_items` e `products`, respeitando as decisões metodológicas estabelecidas nas etapas anteriores.

A construção será realizada de forma modular, separando inicialmente:

- número de pedidos por produto;
- ticket médio por produto;
- status de atividade por produto.

Posteriormente, essas estruturas serão integradas em uma base estatística única e submetidas a validações de consistência.

## Princípio de construção

A base estatística não será criada por meio de uma agregação única sobre todas as tabelas.

Cada variável será construída a partir de sua granularidade apropriada e posteriormente integrada pelo identificador `product_id`.

Dessa forma, será possível validar separadamente:

1. a contagem de pedidos;
2. a receita;
3. o ticket médio;
4. o status de atividade;
5. a correspondência entre maio e junho.

A base final deverá permitir que cada produto da população analítica seja representado por uma única observação, contendo suas métricas correspondentes aos dois períodos.

```python
orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"]
)

print("Pedidos distintos:", orders["order_id"].nunique())
print("Linhas:", len(orders))
print("Duplicidades em order_id:", orders["order_id"].duplicated().sum())

print("\nPeríodo:")
print("Mínimo:", orders["order_purchase_timestamp"].min())
print("Máximo:", orders["order_purchase_timestamp"].max())
```

```python
order_items_path = "../data/raw/olist_order_items_dataset.csv"

order_items = pd.read_csv(order_items_path)

print("Dimensão:", order_items.shape)
print("\nColunas:")
print(order_items.columns.tolist())
```

```python
print("Pedidos distintos:", order_items["order_id"].nunique())
print("Produtos distintos:", order_items["product_id"].nunique())
print("Duplicidades em (order_id, order_item_id):",
      order_items.duplicated(subset=["order_id", "order_item_id"]).sum())

print("\nValores nulos:")
print(order_items[["order_id", "order_item_id", "product_id", "price"]].isna().sum())
```

```python
orders_periodo = orders[
    ["order_id", "order_purchase_timestamp"]
].copy()

print("Dimensão:", orders_periodo.shape)
print("\nNulos:")
print(orders_periodo.isna().sum())
```

```python
items_com_data = order_items.merge(
    orders_periodo,
    on="order_id",
    how="left",
    validate="many_to_one"
)

print("Dimensão:", items_com_data.shape)
print("\nNulos na data de compra:")
print(items_com_data["order_purchase_timestamp"].isna().sum())
```

```python
items_com_data["mes"] = (
    items_com_data["order_purchase_timestamp"]
    .dt.to_period("M")
)

print("Período mínimo:", items_com_data["mes"].min())
print("Período máximo:", items_com_data["mes"].max())

print("\nMeses de interesse:")
print(
    items_com_data[
        items_com_data["mes"].isin(["2018-05", "2018-06"])
    ]["mes"]
    .value_counts()
    .sort_index()
)
```

```python
meses_interesse = [
    pd.Period("2018-05", freq="M"),
    pd.Period("2018-06", freq="M")
]

print(
    items_com_data[
        items_com_data["mes"].isin(meses_interesse)
    ]["mes"]
    .value_counts()
    .sort_index()
)
```

```python
itens_periodo = items_com_data[
    items_com_data["mes"].isin(meses_interesse)
].copy()

pedidos_por_produto = (
    itens_periodo
    .groupby(["product_id", "mes"])["order_id"]
    .nunique()
    .unstack(fill_value=0)
    .rename(columns={
        pd.Period("2018-05", freq="M"): "pedidos_maio",
        pd.Period("2018-06", freq="M"): "pedidos_junho"
    })
    .reset_index()
)

print("Dimensão:", pedidos_por_produto.shape)
print("\nColunas:")
print(pedidos_por_produto.columns.tolist())

print("\nPrimeiras linhas:")
print(pedidos_por_produto.head())
```

```python
products_path = "../data/raw/olist_products_dataset.csv"

products = pd.read_csv(products_path)

print("Dimensão:", products.shape)

print("\nColunas:")
print(products.columns.tolist())
```

```python
products_dim = products[
    ["product_id", "product_category_name"]
].copy()

print("Dimensão:", products_dim.shape)

print("\nProdutos distintos:", products_dim["product_id"].nunique())

print("\nNulos:")
print(products_dim.isna().sum())
```

```python
produtos_com_categoria = pedidos_por_produto.merge(
    products_dim,
    on="product_id",
    how="left",
    validate="one_to_one"
)

print("Dimensão:", produtos_com_categoria.shape)

print("\nNulos na categoria:")
print(
    produtos_com_categoria["product_category_name"].isna().sum()
)

print("\nProdutos distintos:")
print(produtos_com_categoria["product_id"].nunique())
```

```python
orders_periodo = orders[
    ["order_id", "order_purchase_timestamp"]
].copy()

items_com_data = order_items.merge(
    orders_periodo,
    on="order_id",
    how="left",
    validate="many_to_one"
)

print("Dimensão original de order_items:", order_items.shape)
print("Dimensão após o merge:", items_com_data.shape)

print("\nNulos na data de compra:")
print(
    items_com_data["order_purchase_timestamp"].isna().sum()
)
```

```python
items_com_data["mes"] = (
    items_com_data["order_purchase_timestamp"]
    .dt.to_period("M")
)

print("Período inicial:", items_com_data["mes"].min())
print("Período final:", items_com_data["mes"].max())

print("\nQuantidade de itens por mês:")
print(
    items_com_data["mes"]
    .value_counts()
    .sort_index()
)
```

```python
meses_interesse = [
    pd.Period("2018-05", freq="M"),
    pd.Period("2018-06", freq="M")
]

itens_periodo = items_com_data[
    items_com_data["mes"].isin(meses_interesse)
].copy()

print("Dimensão da base maio + junho:", itens_periodo.shape)

print("\nItens por mês:")
print(
    itens_periodo["mes"]
    .value_counts()
    .sort_index()
)
```

```python
pedidos_por_produto = (
    itens_periodo
    .groupby(["product_id", "mes"])["order_id"]
    .nunique()
    .unstack(fill_value=0)
    .rename(columns={
        pd.Period("2018-05", freq="M"): "pedidos_maio",
        pd.Period("2018-06", freq="M"): "pedidos_junho"
    })
    .reset_index()
)

print("Dimensão:", pedidos_por_produto.shape)

print("\nColunas:")
print(pedidos_por_produto.columns.tolist())

print("\nPrimeiras linhas:")
print(pedidos_por_produto.head())
```

```python
print("Produtos com pedidos nos dois meses:",
      (
          (pedidos_por_produto["pedidos_maio"] > 0) &
          (pedidos_por_produto["pedidos_junho"] > 0)
      ).sum())

print("Produtos somente em maio:",
      (
          (pedidos_por_produto["pedidos_maio"] > 0) &
          (pedidos_por_produto["pedidos_junho"] == 0)
      ).sum())

print("Produtos somente em junho:",
      (
          (pedidos_por_produto["pedidos_maio"] == 0) &
          (pedidos_por_produto["pedidos_junho"] > 0)
      ).sum())

print("\nTotal de produtos:",
      pedidos_por_produto["product_id"].nunique())
```

```python
products_dim = products[
    ["product_id", "product_category_name"]
].copy()

produtos_com_categoria = pedidos_por_produto.merge(
    products_dim,
    on="product_id",
    how="left",
    validate="one_to_one"
)

print("Dimensão:", produtos_com_categoria.shape)
print("Produtos distintos:",
      produtos_com_categoria["product_id"].nunique())

print("\nNulos na categoria:",
      produtos_com_categoria["product_category_name"].isna().sum())
```

```python
populacao_produtos = produtos_com_categoria[
    produtos_com_categoria["product_category_name"].isin(categorias_foco)
].copy()

print("Dimensão da população:", populacao_produtos.shape)

print("Produtos distintos:",
      populacao_produtos["product_id"].nunique())

print("Categorias presentes:",
      populacao_produtos["product_category_name"].nunique())
```

```python
distribuicao_categorias = (
    populacao_produtos["product_category_name"]
    .value_counts()
    .sort_values(ascending=False)
)

print(distribuicao_categorias)

print("\nTotal de produtos:",
      distribuicao_categorias.sum())

print("Quantidade de categorias:",
      distribuicao_categorias.size)
```

```python
receita_por_produto = (
    itens_periodo
    .groupby(["product_id", "mes"])["price"]
    .sum()
    .unstack(fill_value=0)
    .rename(columns={
        pd.Period("2018-05", freq="M"): "receita_maio",
        pd.Period("2018-06", freq="M"): "receita_junho"
    })
    .reset_index()
)

print("Dimensão:", receita_por_produto.shape)

print("\nColunas:")
print(receita_por_produto.columns.tolist())

print("\nPrimeiras linhas:")
print(receita_por_produto.head())
```

```python
receita_maio = receita_por_produto["receita_maio"].sum()
receita_junho = receita_por_produto["receita_junho"].sum()

print("Receita total de maio:  R$", round(receita_maio, 2))
print("Receita total de junho: R$", round(receita_junho, 2))
```

```python
base_metricas = pedidos_por_produto.merge(
    receita_por_produto,
    on="product_id",
    how="left",
    validate="one_to_one"
)

print("Dimensão:", base_metricas.shape)

print("Produtos distintos:",
      base_metricas["product_id"].nunique())

print("\nNulos:")
print(
    base_metricas[
        [
            "pedidos_maio",
            "pedidos_junho",
            "receita_maio",
            "receita_junho"
        ]
    ].isna().sum()
)

print("\nPrimeiras linhas:")
print(base_metricas.head())
```

```python
base_estatistica = populacao_produtos.merge(
    receita_por_produto[
        ["product_id", "receita_maio", "receita_junho"]
    ],
    on="product_id",
    how="left",
    validate="one_to_one"
)

print("Dimensão:", base_estatistica.shape)

print("Produtos distintos:",
      base_estatistica["product_id"].nunique())

print("\nNulos:")
print(
    base_estatistica[
        [
            "product_id",
            "product_category_name",
            "pedidos_maio",
            "pedidos_junho",
            "receita_maio",
            "receita_junho"
        ]
    ].isna().sum()
)

print("\nColunas:")
print(base_estatistica.columns.tolist())
```

```python
import numpy as np

base_estatistica["ticket_maio"] = np.where(
    base_estatistica["pedidos_maio"] > 0,
    base_estatistica["receita_maio"] / base_estatistica["pedidos_maio"],
    np.nan
)

base_estatistica["ticket_junho"] = np.where(
    base_estatistica["pedidos_junho"] > 0,
    base_estatistica["receita_junho"] / base_estatistica["pedidos_junho"],
    np.nan
)

print("Ticket maio — valores definidos:",
      base_estatistica["ticket_maio"].notna().sum())

print("Ticket maio — valores indefinidos:",
      base_estatistica["ticket_maio"].isna().sum())

print("\nTicket junho — valores definidos:",
      base_estatistica["ticket_junho"].notna().sum())

print("Ticket junho — valores indefinidos:",
      base_estatistica["ticket_junho"].isna().sum())
```

```python
base_estatistica["atividade_maio"] = (
    base_estatistica["pedidos_maio"] > 0
).astype(int)

base_estatistica["atividade_junho"] = (
    base_estatistica["pedidos_junho"] > 0
).astype(int)

print("Ativos em maio:",
      base_estatistica["atividade_maio"].sum())

print("Inativos em maio:",
      (base_estatistica["atividade_maio"] == 0).sum())

print("\nAtivos em junho:",
      base_estatistica["atividade_junho"].sum())

print("Inativos em junho:",
      (base_estatistica["atividade_junho"] == 0).sum())
```

```python
transicao_atividade = pd.crosstab(
    base_estatistica["atividade_maio"],
    base_estatistica["atividade_junho"]
)

transicao_atividade.index = [
    "Inativo em maio",
    "Ativo em maio"
]

transicao_atividade.columns = [
    "Inativo em junho",
    "Ativo em junho"
]

print(transicao_atividade)
```

```python
print("DIMENSÃO")
print(base_estatistica.shape)

print("\nUNICIDADE")
print("Produtos distintos:", base_estatistica["product_id"].nunique())

print("\nNULOS")
print(base_estatistica.isna().sum())

print("\nRECEITAS NEGATIVAS")
print(
    (base_estatistica[["receita_maio", "receita_junho"]] < 0).sum()
)

print("\nINCONSISTÊNCIAS DE ATIVIDADE")

inconsistencia_maio = (
    base_estatistica["atividade_maio"]
    != (base_estatistica["pedidos_maio"] > 0).astype(int)
).sum()

inconsistencia_junho = (
    base_estatistica["atividade_junho"]
    != (base_estatistica["pedidos_junho"] > 0).astype(int)
).sum()

print("Maio:", inconsistencia_maio)
print("Junho:", inconsistencia_junho)

print("\nTICKET SEM PEDIDO")

ticket_maio_invalido = (
    (base_estatistica["pedidos_maio"] == 0) &
    (base_estatistica["ticket_maio"].notna())
).sum()

ticket_junho_invalido = (
    (base_estatistica["pedidos_junho"] == 0) &
    (base_estatistica["ticket_junho"].notna())
).sum()

print("Maio:", ticket_maio_invalido)
print("Junho:", ticket_junho_invalido)
```

# Parte 4 — Formulação das Hipóteses

## Objetivo

Transformar os principais padrões identificados durante a Análise Exploratória de Dados (EDA) em hipóteses estatísticas formalmente testáveis.

Nesta etapa serão definidas:

- as perguntas estatísticas;
- as variáveis envolvidas;
- a unidade estatística;
- as hipóteses nula (H₀) e alternativa (H₁);
- a direção da investigação, quando aplicável;
- os critérios que posteriormente orientarão a escolha dos testes estatísticos.

A aplicação dos testes não será realizada nesta etapa.

## Princípio metodológico

A escolha do método estatístico será realizada somente após a formulação das hipóteses e o diagnóstico da estrutura das variáveis.

Portanto, a sequência adotada será:

**hipótese → diagnóstico dos dados → escolha do método → teste estatístico.**

## Hipóteses investigadas

A análise estatística será organizada em três dimensões principais:

1. alteração no número de pedidos por produto;
2. alteração no ticket médio dos produtos ativos nos dois períodos;
3. alteração no status de atividade dos produtos.

Essas dimensões representam aspectos distintos do desempenho comercial e serão analisadas separadamente.

## População inferencial

A população principal da análise é composta por 2.413 produtos pertencentes às dez categorias prioritárias identificadas durante a EDA.

O grupo de 104 produtos utilizado na análise exploratória de concentração das perdas permanecerá como grupo descritivo e diagnóstico, não sendo utilizado como população inferencial para testar sua própria seleção.

## Unidade estatística

A unidade estatística permanece sendo o `product_id`.

Cada produto possui observações correspondentes aos períodos de maio e junho de 2018, permitindo a comparação entre os dois períodos dentro da mesma unidade.

## 4.2 — Hipótese 1: Número de pedidos

### Pergunta estatística

A quantidade de pedidos por produto apresentou alteração sistemática entre maio e junho de 2018 na população dos 2.413 produtos analisados?

### Variável

`pedidos_maio` e `pedidos_junho`.

A variável representa a quantidade de pedidos distintos nos quais determinado produto apareceu em cada período.

### Unidade estatística

`product_id`.

Cada produto possui uma observação em maio e outra em junho, caracterizando uma estrutura pareada.

### Diferença entre os períodos

Para cada produto, a alteração poderá ser representada por:

\[
\Delta Pedidos_i =
Pedidos_{i,junho} - Pedidos_{i,maio}
\]

Valores negativos representam redução de pedidos para o produto, enquanto valores positivos representam aumento.

### Hipótese nula (H₀)

Não há evidência de alteração sistemática na quantidade de pedidos por produto entre maio e junho de 2018.

### Hipótese alternativa (H₁)

Há evidência de alteração sistemática na quantidade de pedidos por produto entre maio e junho de 2018.

### Observação metodológica

A hipótese será inicialmente formulada de maneira bilateral.

A escolha do teste estatístico será realizada posteriormente, após o diagnóstico da distribuição da variável, das diferenças entre os períodos, da presença de zeros e das demais características relevantes da estrutura dos dados.

Nenhuma hipótese de causalidade será testada nesta etapa.

## 4.3 — Hipótese 2: Ticket médio

### Pergunta estatística

Entre os produtos que apresentaram pedidos tanto em maio quanto em junho de 2018, houve alteração sistemática no ticket médio entre os dois períodos?

### Variável

`ticket_maio` e `ticket_junho`.

O ticket representa a receita de produtos dividida pela quantidade de pedidos distintos nos quais o produto apareceu no respectivo período.

\[
Ticket_{i,mês} =
\frac{Receita_{i,mês}}
{Pedidos_{i,mês}}
\]

### População analisada

A comparação será realizada somente entre os produtos que apresentaram pelo menos um pedido em ambos os períodos:

- `pedidos_maio > 0`;
- `pedidos_junho > 0`.

Essa condição corresponde aos 399 produtos classificados como ativos nos dois meses.

### Justificativa

Produtos sem pedidos em determinado mês não possuem ticket definido naquele período. Portanto, produtos que passaram de ativo para inativo ou de inativo para ativo não podem ser utilizados diretamente em uma comparação pareada de ticket.

A exclusão desses produtos desta hipótese não representa exclusão da população geral da análise. Eles permanecem considerados nas análises de pedidos e atividade.

### Diferença entre os períodos

Para cada produto ativo nos dois meses:

\[
\Delta Ticket_i =
Ticket_{i,junho} - Ticket_{i,maio}
\]

Valores negativos representam redução do ticket para o produto, enquanto valores positivos representam aumento.

### Hipótese nula (H₀)

Não há evidência de alteração sistemática no ticket médio por produto entre maio e junho de 2018 entre os produtos ativos nos dois períodos.

### Hipótese alternativa (H₁)

Há evidência de alteração sistemática no ticket médio por produto entre maio e junho de 2018 entre os produtos ativos nos dois períodos.

### Observação metodológica

A hipótese será inicialmente formulada de maneira bilateral.

A escolha do teste estatístico será realizada posteriormente, após o diagnóstico da distribuição dos tickets e das diferenças entre os períodos.

A análise do ticket não será interpretada isoladamente como explicação causal para a alteração da receita.

## 4.4 — Hipótese 3: Atividade dos produtos

### Pergunta estatística

A proporção de produtos ativos apresentou alteração entre maio e junho de 2018?

### Variável

`atividade_maio` e `atividade_junho`.

A variável é binária:

- `1` = produto apresentou pelo menos um pedido no período;
- `0` = produto não apresentou pedidos no período.

### Unidade estatística

`product_id`.

Os mesmos 2.413 produtos são observados nos dois períodos, caracterizando uma estrutura pareada.

### Estrutura de transição

A comparação entre os períodos pode ser representada por quatro estados:

- inativo → inativo;
- inativo → ativo;
- ativo → inativo;
- ativo → ativo.

Na população analisada foram observadas:

- 0 produtos inativo → inativo;
- 934 produtos inativo → ativo;
- 1.080 produtos ativo → inativo;
- 399 produtos ativo → ativo.

### Hipótese nula (H₀)

Não há evidência de alteração sistemática na proporção de produtos ativos entre maio e junho de 2018.

### Hipótese alternativa (H₁)

Há evidência de alteração sistemática na proporção de produtos ativos entre maio e junho de 2018.

### Observação metodológica

A hipótese será inicialmente formulada de maneira bilateral.

A escolha do procedimento estatístico será realizada posteriormente, considerando a natureza binária e pareada da variável e a estrutura observada na matriz de transição.

A alteração da atividade será interpretada como evidência descritiva/inferencial de mudança de estado, não como evidência de causa da redução da receita.

## 4.5 — Matriz Consolidada das Hipóteses

A análise estatística foi estruturada em três hipóteses principais, correspondentes a dimensões distintas do desempenho comercial.

| Hipótese | Pergunta estatística | Variável principal | População | Estrutura dos dados | Método estatístico |
|---|---|---|---|---|---|
| H1 — Pedidos | A quantidade de pedidos por produto apresentou alteração sistemática entre maio e junho de 2018? | `pedidos_maio` / `pedidos_junho` | 2.413 produtos | Quantitativa discreta e pareada | A definir após diagnóstico |

| H3 — Atividade | A proporção de produtos ativos apresentou alteração entre maio e junho de 2018? | `atividade_maio` / `atividade_junho` | 2.413 produtos | Binária e pareada | A definir após diagnóstico |

### Síntese metodológica

As três hipóteses possuem a mesma unidade estatística (`product_id`) e comparam os mesmos períodos, porém apresentam estruturas de dados distintas.

### Regra metodológica

Nenhum teste estatístico será escolhido exclusivamente com base no tipo nominal da variável ou por uma regra automática.

A escolha será justificada pela combinação entre:

1. pergunta estatística;
2. unidade de análise;
3. estrutura pareada;
4. natureza da variável;
5. distribuição observada;
6. pressupostos do método;

## 4.6 — Critérios para Escolha do Método Estatístico

A escolha dos procedimentos estatísticos será realizada somente após o diagnóstico das variáveis e da estrutura das diferenças entre maio e junho de 2018.

O método não será definido exclusivamente pelo tipo da variável ou por uma regra automática. Serão considerados conjuntamente a pergunta estatística, a unidade de análise, a estrutura pareada, a distribuição dos dados e os pressupostos dos métodos candidatos.

### Critérios gerais

### Hipótese 1 — Pedidos

Como se trata de uma variável de contagem, serão considerados aspectos como:

### Princípio de decisão

A escolha do teste será feita antes da interpretação do resultado inferencial e será justificada com base nas características observadas nos dados.

# Parte 5 — Diagnóstico das Variáveis e Pressupostos

## Objetivo

Avaliar a estrutura das variáveis utilizadas nas três hipóteses estatísticas antes da aplicação dos testes inferenciais.

O diagnóstico buscará identificar características dos dados que possam influenciar a escolha dos procedimentos estatísticos, incluindo:

- distribuição das variáveis;
- distribuição das diferenças entre maio e junho;
- presença e proporção de zeros;
- assimetria;
- valores extremos;
- escala e dispersão;
- estrutura dos pares;
- características específicas das variáveis quantitativas e binárias.

## Princípio metodológico

O diagnóstico não será utilizado para selecionar artificialmente um método que produza determinado resultado.

Seu objetivo é verificar quais características dos dados são relevantes para avaliar a adequação dos métodos estatísticos candidatos.

A escolha definitiva dos testes será realizada somente após a conclusão dos diagnósticos correspondentes a cada hipótese.

## Ordem do diagnóstico

A análise será realizada separadamente para cada hipótese:

1. pedidos;
2. ticket;

## 5.4 — Decisão Metodológica para H1

### Método selecionado

Para a hipótese relacionada à alteração do número de pedidos, a análise inferencial principal será baseada na **direção das diferenças pareadas**, utilizando o teste binomial exato, equivalente à abordagem do teste dos sinais para dois estados possíveis.

### Justificativa

A variável `delta_pedidos` apresentou características que tornam inadequada uma escolha automática baseada na média ou em pressupostos de distribuição aproximadamente simétrica:

- natureza discreta e de contagem;
- forte concentração das diferenças em `-1`, `0` e `+1`;
- 148 diferenças iguais a zero;
- mediana igual a `-1`;
- IQR igual a `2`;
- assimetria igual a aproximadamente `-14,65`;
- presença de valores extremos, com mínimo igual a `-98`;
- apenas 1% das observações apresentou magnitude absoluta superior a aproximadamente 9 pedidos.

Diante dessa estrutura, a análise inferencial principal será direcionada à **proporção de reduções entre os produtos que apresentaram alteração**, evitando que a conclusão principal dependa diretamente de poucos valores extremos.

### Hipótese estatística operacional

Considerando somente as diferenças não nulas:

\[
H_0: p = 0,5
\]

\[
H_1: p \neq 0,5
\]

onde `p` representa a probabilidade de um produto apresentar redução de pedidos entre maio e junho, condicionada à ocorrência de alteração no número de pedidos.

### Estatística observada

Entre os 2.265 produtos com diferenças não nulas:

- 1.224 apresentaram redução;
- 1.041 apresentaram aumento;
- proporção observada de reduções: 54,04%;
- proporção observada de aumentos: 45,96%;
- diferença entre as proporções: 8,08 pontos percentuais.

### Observação metodológica

Os 148 produtos com `delta_pedidos = 0` não possuem direção de mudança e, portanto, não participam diretamente da comparação entre reduções e aumentos.

A exclusão dos zeros é específica desta hipótese operacional e não representa exclusão desses produtos da população geral da análise.

A magnitude das alterações continuará sendo analisada separadamente por meio de estatísticas descritivas e medidas de efeito, evitando interpretar a análise de direção como se ela representasse a magnitude da variação dos pedidos.

```python
from scipy.stats import binomtest

resultado_h1 = binomtest(
    k=negativas,
    n=total_nao_nulas,
    p=0.5,
    alternative="two-sided"
)

print("Sucessos (reduções):", negativas)
print("Total de diferenças não nulas:", total_nao_nulas)
print("Proporção observada:", resultado_h1.statistic)
print("p-valor:", resultado_h1.pvalue)
```

```python
intervalo_h1 = resultado_h1.proportion_ci(
    confidence_level=0.95,
    method="exact"
)

print("Proporção observada:", resultado_h1.statistic)
print("IC 95% — limite inferior:", intervalo_h1.low)
print("IC 95% — limite superior:", intervalo_h1.high)
```

## 6.5 — Resultado Inferencial da H1

O teste binomial exato bilateral foi utilizado para avaliar se, entre os produtos que apresentaram alteração no número de pedidos, a proporção de reduções diferia de 50%.

Foram observados:

- 1.224 produtos com redução;
- 1.041 produtos com aumento;
- 2.265 produtos com diferença não nula;
- proporção observada de reduções: 54,04%;
- p-valor: 0,0001302;
- intervalo de confiança de 95% para a proporção de reduções: 51,96% a 56,11%.

Como o p-valor é inferior ao nível de significância de 5% e o intervalo de confiança não contém a proporção de referência de 50%, rejeita-se a hipótese nula de equilíbrio entre reduções e aumentos.

Os resultados fornecem evidência estatística de que, entre os produtos que apresentaram alteração no número de pedidos, a proporção de reduções foi diferente de 50%, apresentando estimativa de 54,04%.

Essa evidência descreve a direção das alterações observadas entre os produtos e não deve ser interpretada como evidência de causalidade ou como medida direta da magnitude da redução da receita operacional.

```python
proporcao_observada = resultado_h1.statistic
proporcao_referencia = 0.5

diferenca_proporcao = (
    proporcao_observada - proporcao_referencia
)

print("Proporção observada:", proporcao_observada)
print("Proporção de referência:", proporcao_referencia)
print("Diferença:", diferenca_proporcao)
print(
    "Diferença em pontos percentuais:",
    diferenca_proporcao * 100
)
```

```python
delta_nao_nulo = base_h2.loc[
    base_h2["delta_ticket_teste"] != 0,
    "delta_ticket_teste"
]

postos = delta_nao_nulo.abs().rank(method="average")

soma_postos_positivos = postos[
    delta_nao_nulo > 0
].sum()

soma_postos_negativos = postos[
    delta_nao_nulo < 0
].sum()

soma_total_postos = postos.sum()

rank_biserial = (
    soma_postos_positivos - soma_postos_negativos
) / soma_total_postos

print("Soma dos postos positivos:", soma_postos_positivos)
print("Soma dos postos negativos:", soma_postos_negativos)
print("Soma total dos postos:", soma_total_postos)
print("Rank-biserial correlation:", rank_biserial)
```

```python
import numpy as np

delta_h2_bootstrap = delta_nao_nulo.to_numpy()

rng = np.random.default_rng(42)

n_bootstrap = 10000
efeitos_bootstrap = []

for _ in range(n_bootstrap):
    amostra = rng.choice(
        delta_h2_bootstrap,
        size=len(delta_h2_bootstrap),
        replace=True
    )

    postos_boot = pd.Series(np.abs(amostra)).rank(method="average")

    soma_positivos = postos_boot[amostra > 0].sum()
    soma_negativos = postos_boot[amostra < 0].sum()

    efeito = (
        soma_positivos - soma_negativos
    ) / postos_boot.sum()

    efeitos_bootstrap.append(efeito)

efeitos_bootstrap = np.array(efeitos_bootstrap)

print("Bootstrap concluído.")
print("Número de reamostragens:", len(efeitos_bootstrap))
```

```python
ic_inferior = np.percentile(efeitos_bootstrap, 2.5)
ic_superior = np.percentile(efeitos_bootstrap, 97.5)

print("Rank-biserial observado:", rank_biserial)
print("IC 95% — limite inferior:", ic_inferior)
print("IC 95% — limite superior:", ic_superior)
```

## 6.2 — Síntese interpretativa das evidências

A análise estatística avaliou três possíveis mecanismos relacionados à deterioração observada entre maio e junho de 2018: mudança no volume de pedidos dos produtos, mudança no ticket e mudança na atividade dos produtos.

### H1 — Pedidos

A H1 foi sustentada pelo teste binomial exato (**p = 0,00013**). Entre os produtos que apresentaram mudança não nula no número de pedidos, **54,04% apresentaram redução**, contra 45,96% que apresentaram aumento.

A diferença em relação ao equilíbrio foi de **4,04 pontos percentuais**, com intervalo de confiança de 95% para a proporção de reduções entre **51,96% e 56,11%**.

O resultado indica uma assimetria estatisticamente significativa na direção das mudanças de pedidos, com maior frequência de reduções do que de aumentos.

Esse resultado, entretanto, não mede a magnitude da perda de receita nem estabelece que a redução de pedidos seja causada por um determinado fator operacional.

### H2 — Ticket

A H2 não foi sustentada pelo teste de Wilcoxon signed-rank (**p = 0,85034**).

Entre os 399 produtos que permaneceram ativos nos dois meses, não foi identificada evidência de uma mudança sistemática no ticket.

O efeito rank-biserial foi de aproximadamente **−0,016**, com intervalo de confiança de 95% entre **−0,184 e 0,149**, abrangendo o valor nulo.

Embora alguns produtos tenham apresentado variações individuais de ticket, os resultados não sustentam a existência de uma mudança generalizada nessa dimensão.

### H3 — Atividade

A H3 foi sustentada pelo teste exato de McNemar (**p = 0,00123**).

Entre os produtos que apresentaram mudança de estado de atividade, foram observadas **1.080 saídas (`1 → 0`) contra 934 entradas (`0 → 1`)**, resultando em saldo líquido de **146 produtos ativos a menos**.

Entre as 2.014 transições discordantes, as saídas representaram **53,62%**, correspondendo a uma assimetria de **7,25 pontos percentuais**.

O intervalo de confiança de 95% para essa assimetria foi de **2,84 a 11,64 pontos percentuais**, indicando que o desequilíbrio entre saídas e entradas é estatisticamente diferente de zero.

### Síntese conjunta

Considerados em conjunto, os resultados fornecem evidência de mudanças sistemáticas relacionadas ao **volume de pedidos e à atividade dos produtos**, enquanto não fornecem evidência de uma mudança sistemática no **ticket** entre os produtos ativos nos dois períodos.

Dessa forma, a evidência estatística direciona a investigação para mecanismos relacionados à **redução de volume e de atividade comercial**, em vez de uma deterioração generalizada do valor por pedido.

Essa interpretação ainda não constitui uma conclusão causal. Os testes identificam padrões estatisticamente sustentados, mas não determinam quais fatores operacionais, comerciais ou relacionados aos produtos produziram essas mudanças.
