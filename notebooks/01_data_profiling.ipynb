```python
from pathlib import Path

RAW_DIR = Path("../data/raw")

arquivos = sorted(RAW_DIR.glob("*.csv"))

for arquivo in arquivos:
    print(arquivo.name)
```

```python
import pandas as pd

tabelas = {}

for arquivo in arquivos:
    df = pd.read_csv(arquivo)
    tabelas[arquivo.stem] = df
```

```python
print(f"Quantidade de tabelas carregadas: {len(tabelas)}")

for nome, df in tabelas.items():
    print(f"{nome}: {df.shape}")
```

```python
# Seleciona a tabela de clientes
df_customers = tabelas["olist_customers_dataset"]

# Mostra informações básicas da tabela
print("=" * 70)
print("TABELA: olist_customers_dataset")
print("=" * 70)

print(f"Quantidade de linhas: {df_customers.shape[0]}")
print(f"Quantidade de colunas: {df_customers.shape[1]}")

print("\nColunas da tabela:")

for numero, coluna in enumerate(df_customers.columns, start=1):
    print(f"{numero}. {coluna}")
```
