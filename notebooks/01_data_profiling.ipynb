```python
totais_items = (
    df_items
    .groupby("order_id")
    .agg(
        valor_produtos=("price", "sum"),
        valor_frete=("freight_value", "sum")
    )
    .reset_index()
)

totais_items["valor_itens_frete"] = (
    totais_items["valor_produtos"] +
    totais_items["valor_frete"]
)

totais_items.head()
```

```python
totais_payments = (
    df_payments
    .groupby("order_id")
    .agg(
        valor_pagamentos=("payment_value", "sum")
    )
    .reset_index()
)

totais_payments.head()
```

```python
comparacao_valores = totais_items.merge(
    totais_payments,
    on="order_id",
    how="inner"
)

comparacao_valores.head()
```

```python
comparacao_valores["diferenca"] = (
    comparacao_valores["valor_pagamentos"]
    - comparacao_valores["valor_itens_frete"]
)

comparacao_valores["diferenca"].describe()
```

```python
import numpy as np

comparacao_valores["corresponde"] = np.isclose(
    comparacao_valores["valor_pagamentos"],
    comparacao_valores["valor_itens_frete"],
    atol=0.01
)

comparacao_valores["corresponde"].value_counts()
```

```python
divergencias = (
    comparacao_valores[
        ~comparacao_valores["corresponde"]
    ]
    .copy()
    .sort_values("diferenca")
)

divergencias[
    [
        "order_id",
        "valor_produtos",
        "valor_frete",
        "valor_itens_frete",
        "valor_pagamentos",
        "diferenca"
    ]
].head(10)
```

```python
comparacao_valores["diferenca_abs"] = (
    comparacao_valores["diferenca"].abs()
)

comparacao_valores["categoria_divergencia"] = np.select(
    [
        comparacao_valores["diferenca_abs"] <= 0.05,
        comparacao_valores["diferenca_abs"] <= 10,
        comparacao_valores["diferenca_abs"] <= 50,
        comparacao_valores["diferenca_abs"] > 50
    ],
    [
        "Até R$ 0,05",
        "R$ 0,06 a R$ 10",
        "R$ 10,01 a R$ 50",
        "Acima de R$ 50"
    ],
    default="Sem divergência"
)

comparacao_valores["categoria_divergencia"].value_counts()
```

```python
analise_pagamento = (
    comparacao_valores
    .merge(
        df_payments[
            ["order_id", "payment_type"]
        ].drop_duplicates("order_id"),
        on="order_id",
        how="left"
    )
    .groupby("payment_type")
    .agg(
        total_pedidos=("order_id", "count"),
        pedidos_com_divergencia=("corresponde", lambda x: (~x).sum())
    )
    .reset_index()
)

analise_pagamento["taxa_divergencia"] = (
    analise_pagamento["pedidos_com_divergencia"]
    / analise_pagamento["total_pedidos"]
    * 100
)

analise_pagamento.sort_values(
    "taxa_divergencia",
    ascending=False
)
```

A tabela order_payments possui granularidade de registro de pagamento, permitindo múltiplos registros por pedido. Embora a maior parte dos pedidos apresente compatibilidade entre o valor registrado nos pagamentos e a soma de produtos e frete, foram identificadas 300 divergências entre 98.665 pedidos comparáveis. As divergências estão distribuídas ao longo do período analisado e apresentam baixa frequência relativa entre os diferentes tipos de pagamento. Dessa forma, payment_value não será utilizado como fonte primária para o cálculo da receita de produtos no CASE 01. A receita de produtos será calculada a partir de order_items.price, enquanto payment_value será tratado como uma variável relacionada ao comportamento de pagamento.

# Consolidação do Data Profiling

## Principais achados

A análise do desempenho mensal mostrou que a receita apresenta comportamento oscilatório ao longo do período observado, não sendo identificada uma trajetória de queda contínua.

Considerando os meses completos e o período mais recente de maio para junho de 2018, observou-se:

- redução de 13,19% na receita;
- redução de 10,11% no número de pedidos;
- redução de 3,42% no ticket médio.

A decomposição da receita foi validada matematicamente por meio da relação:

**Receita = Pedidos × Ticket médio**

A validação foi realizada tanto para o desempenho agregado quanto para as categorias prioritárias analisadas.

## Comportamento por categoria

A análise das categorias mostrou que a deterioração recente não ocorreu de maneira uniforme.

Entre as categorias selecionadas pelas maiores perdas absolutas de receita entre maio e junho de 2018:

- `relogios_presentes` apresentou a maior redução absoluta, de R$ 36.987,10;
- `ferramentas_jardim` apresentou redução de R$ 19.844,47;
- `esporte_lazer` apresentou redução de R$ 13.813,10;
- `moveis_decoracao` apresentou redução de R$ 12.070,76;
- `cool_stuff` apresentou redução de R$ 10.722,74.

A decomposição por pedidos e ticket mostrou mecanismos diferentes entre as categorias. Em algumas delas, a redução da receita ocorreu acompanhada por queda simultânea no número de pedidos e no ticket médio. Em outras, o ticket médio aumentou enquanto o número de pedidos caiu.

Esse comportamento indica que a deterioração da receita não deve ser analisada exclusivamente pelo ticket médio ou exclusivamente pelo volume de pedidos.

## Compensação entre categorias

As dez categorias selecionadas apresentaram, em conjunto, redução de R$ 135.440,90 entre maio e junho de 2018, enquanto a queda líquida da receita total da operação foi de R$ 131.393,37.

A diferença indica que outras categorias apresentaram crescimento no mesmo período, compensando parcialmente parte das perdas observadas nas categorias selecionadas.

Portanto, a deterioração recente não foi uniforme em todo o portfólio de categorias.

## Limitações e cuidados de interpretação

Os meses de setembro de 2016 e setembro de 2018 possuem cobertura parcial e não foram utilizados na análise comparativa principal.

A comparação de maio para junho de 2018 representa um recorte recente de deterioração e não deve ser interpretada, isoladamente, como evidência de uma tendência estrutural de queda.

Além disso, grandes variações percentuais podem ocorrer em categorias com baixo volume de vendas. Por esse motivo, a priorização considerou conjuntamente variação percentual, impacto financeiro absoluto e comportamento de pedidos e ticket.

Os resultados desta etapa são descritivos. Não estabelecem causalidade.

## Hipóteses para a etapa de EDA

Com base nos achados do Data Profiling, a etapa seguinte deverá investigar:

1. se a deterioração recente está concentrada em determinados produtos dentro das categorias prioritárias;
2. se a redução do número de pedidos está concentrada em determinados produtos ou segmentos;
3. se determinados sellers apresentam comportamento associado à redução observada;
4. se a redução da receita está relacionada principalmente à quantidade de itens vendidos, ao ticket médio ou à combinação desses fatores em diferentes segmentos;
5. se existem padrões temporais ou de concentração que ajudem a explicar a deterioração observada.

Essas hipóteses serão investigadas na etapa de EDA e não constituem conclusões causais do Data Profiling.
